```python
import gymnasium as gym
import numpy as np
import random
```

```python
# sample env
env = gym.make('CliffWalking-v1')
```

```python
print(env.observation_space.n) #states
print(env.action_space.n) # actions

starting_state, _ = env.reset() # 36 = 3 * 12 + 0
print(starting_state) #36
```

#SARSA Implementation

```python
gamma = 0.99
alpha = 0.5
epsilon = 0.1
episodes = 500
```

```python
# Q - table => store Q-values
Q = np.zeros((env.observation_space.n, env.action_space.n))
```

```python
# Policy - epsilon-greedy : state -> action

def epsilon_greedy(state):
    if random.random() < epsilon:
        return env.action_space.sample() # random action => EXPLORE
    else:
        return np.argmax(Q[state]) # greedy action => EXPLOIT
```

```python
for episode in range(episodes):

    env = gym.make("CliffWalking-v1")

    done = False # when the episode has ended
    state, _ = env.reset()
    action = epsilon_greedy(state)

    total_reward = 0
    episodes_len = 0

    while not done:
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated
        next_action = epsilon_greedy(next_state)

        # SARSA update
        Q[state, action] += alpha * (reward + gamma * Q[next_state, next_action] - Q[state, action])

        state = next_state
        action = next_action

        total_reward += reward
        episodes_len += 1

    print(f"Episode: {episode + 1}/500, Reward: {total_reward}, Episode Length: {episodes_len})")
    env.close()
```

```python
# what did the agent learn

env = gym.make('CliffWalking-v1', render_mode='human')
state, _ = env.reset()
done = False
total_reward = 0
episode_len = 0

while not done:
    action = np.argmax(Q[state])
    next_state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated

    state = next_state
    total_reward += reward
    episode_len += 1

print(f"Total Reward: {total_reward}, Episode Length: {episode_len}")
env.close()
```
