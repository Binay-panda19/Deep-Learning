#Resizing the Images

# Convert RGB Images into Grayscale using OpenCV

```python
# importing openCV
import cv2
```

```python
img = cv2.imread('/content/YjJqdGZqMDguanBn')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
type(grayscale_img)
```

```python
grayscale_img.shape
```
